```python
import fitz
import spacy
import re
import json

from google.colab import files
```

# Load spaCy Model

```python
nlp = spacy.load("en_core_web_lg")

print("spaCy Model Loaded Successfully ✅")
```

# Upload Resume

```python
uploaded = files.upload()
```

# Extract Text

```python
filename = next(iter(uploaded))

pdf = fitz.open(filename)

resume_text = ""

for page in pdf:
  resume_text+=page.get_text()

print(resume_text)
```

# Process Resume

```python
doc = nlp(resume_text)
```

# Extract Name

```python
name = None

for ent in doc.ents:
  if ent.label_=="PERSON":
    name = ent.text
    break


print(name)
```

# Extract Email

```python
email_pattern = r'[\w\.-]+@[\w\.-]+\.\w+'

emails = re.findall(email_pattern, resume_text)

emails
```

# Extract Phone Number

```python
phone_pattern = r'(\+?\d[\d\s\-\(\)]{8,}\d)'

phones = re.findall(phone_pattern, resume_text)

phones
```

# Add EntityRuler

```python
from spacy.pipeline import EntityRuler

if "entity_ruler" not in nlp.pipe_names:
  ruler = nlp.add_pipe("entity_ruler", before="ner")
else:
  ruler = nlp.get_pipe("entity_ruler")
```

# Skill Patterns

```python
patterns = [

    {"label":"SKILL","pattern":"Python"},
    {"label":"SKILL","pattern":"Java"},
    {"label":"SKILL","pattern":"JavaScript"},
    {"label":"SKILL","pattern":"React"},
    {"label":"SKILL","pattern":"Next.js"},
    {"label":"SKILL","pattern":"Node.js"},
    {"label":"SKILL","pattern":"Express"},
    {"label":"SKILL","pattern":"MongoDB"},
    {"label":"SKILL","pattern":"PostgreSQL"},
    {"label":"SKILL","pattern":"MySQL"},
    {"label":"SKILL","pattern":"Redis"},
    {"label":"SKILL","pattern":"Docker"},
    {"label":"SKILL","pattern":"Kubernetes"},
    {"label":"SKILL","pattern":"AWS"},
    {"label":"SKILL","pattern":"Git"},
    {"label":"SKILL","pattern":"GitHub"},
    {"label":"SKILL","pattern":"FastAPI"},
    {"label":"SKILL","pattern":"Spring Boot"},
    {"label":"SKILL","pattern":"TensorFlow"},
    {"label":"SKILL","pattern":"PyTorch"},
    {"label":"SKILL","pattern":"Machine Learning"},
    {"label":"SKILL","pattern":"Deep Learning"},
    {"label":"SKILL","pattern":"NLP"},
    {"label":"SKILL","pattern":"scikit-learn"}
]

ruler.add_patterns(patterns)
```

# Detect Skills

```python
doc = nlp(resume_text)

skills = []

for ent in doc.ents:

    if ent.label_ == "SKILL":
        skills.append(ent.text)

skills = sorted(list(set(skills)))

skills
```

# Extract Education

```python
education_keywords = [
    "B.Tech",
    "Bachelor",
    "Master",
    "M.Tech",
    "B.E",
    "BSc",
    "MSc",
    "PhD",
    "Diploma"
]

education = []

for keyword in education_keywords:

    if keyword.lower() in resume_text.lower():
        education.append(keyword)

education
```

# Extract Experience

```python
experience_keywords = [

    "Intern",
    "Internship",
    "Software Engineer",
    "Developer",
    "Full Stack",
    "Backend",
    "Frontend"

]

experience = []

for keyword in experience_keywords:

    if keyword.lower() in resume_text.lower():
        experience.append(keyword)

experience
```

# Create JSON

```python
resume_json = {

    "name": name,

    "email": emails[0] if emails else None,

    "phone": phones[0] if phones else None,

    "skills": skills,

    "education": education,

    "experience": experience

}
```

# Pretty Print

```python
print(json.dumps(resume_json, indent=4))
```
